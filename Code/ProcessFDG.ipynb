```python
base = "/Users/SeanKelly1/Documents/FinalThesis(2023Ulster)/"
save_folder = "/Users/SeanKelly1/Documents/FinalThesis(2023Ulster)/Raw/temp_fdg/"
processed_folder = "/Users/SeanKelly1/Documents/FinalThesis(2023Ulster)/Raw/All/FDG"
os.chdir(save_folder)
file_fdg = os.listdir()
n = 0
 
for fdg in file_fdg:
    temp_img = []
    if fdg == ".DS_Store":
        pass
    else:
        tensor_nifti = nb.load(fdg) # The pics is extracted but remains in a nibabel object
        temp_arr = tensor_nifti.get_fdata() # Converts the nibabel object to numpy array
        if temp_arr.shape[3] < 51 or temp_arr.shape[2] < 63:
            pass
        else:
            subject_id = fdg.split("-")[1].split("_")[0]
            assessement_id = fdg.split("-")[2].split("_")[0]
            typ = fdg.split("-")[-1].split("_")[0]
            raw_df = pd.read_csv(base + "Spreadsheet/ADRC Clinical Data.csv")
            cdr = max(raw_df[raw_df["Subject"] == subject_id]["cdr"])
            new_file_name = "{}_{}_{}".format(subject_id, typ, cdr)
            # os.listdir(save_folder)
            for i in range(42, 51):
                for y in range(18, 41):
                    temp = temp_arr[:, :, y, i]
                    new_file_name = "{}_{}_{}_{}_{}".format(subject_id, typ, cdr, y, i)
                    temp_folder = (processed_folder + "/" + str(cdr) + "/" + new_file_name + ".png")
                    plt.imsave(temp_folder, temp)
                    n += 1
    print("{} has been processed".format(fdg))
                
print("{} images saved and classified by cdr".format(n))

# 25047 images was saved and classified by cdr. There was no rating of 3.0 in the FDG file.
```
