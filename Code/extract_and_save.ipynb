# Extract and Save

### This file includes code for extracting NIFTI files from sub folders, converting respective slice to jpeg files.

```python
# Imports 

import os 
import cv2
import nibabel as nb
import numpy as np
import pandas as pd
import torch as to
import torchvision
import torchvision.transforms as T
from PIL import Image
import matplotlib.pyplot as plt
```

```python
parent_dir = "/Users/SeanKelly1/Documents/Final Thesis (2023 Ulster)/Raw/MR_nifti"            # Define the parent directory where the parent folder is located.
```

```python
os.chdir(parent_dir)

savefolder = "/Users/SeanKelly1/Documents/Final Thesis (2023 Ulster)/Raw/MR_nifti_images_01/"

if not os.path.exists(savefolder):
    os.makedirs(savefolder)
    print("Save folder created")
else:
    print("Save folder exists")

all_nifti = os.listdir()
Typ = {}
initial = 0 

for each_nifti in all_nifti:
    try:
        nifti_filename = str(each_nifti)
        subject_id = nifti_filename.split("-")[1].split("_")[0]
        assessement_id = nifti_filename.split("-")[2].split("_")[0]

        typ = nifti_filename.split("-")[-1].split("_")[-1].split(".")[0]
        typ = typ + str(initial)

        
        lookup_name = subject_id + "_ClinicalData_" + assessement_id
        raw_df = pd.read_csv("../../Spreadsheet/ADRC Clinical Data.csv")
        
        cdr = max(raw_df[raw_df["Subject"] == subject_id]["cdr"])
        ### Quick note on above line of code:
        ### A look up into the clinical data is done using the subject ID as some of the experiments have not been included in the published spreadsheet.
        ### Since there may be several assessements from one subject, the subject cdr is then queried and the maximum cdr is selected. This gives a sense of the developemnt of dementia in the subject.
        ### This furthermore helps teach the model to identify trends which may not be recognisable to humans as this is one of the objectives of the research.

        pic_name = subject_id + "_" + assessement_id + "_" + str(cdr)+ "_"

        pics_array = nb.load(each_nifti) # The pics is extracted but remains in a nibabel object
        
        pics_np_array = pics_array.get_fdata() # Converts the nibabel object to numpy array

        # pics_np_array.shape # The array needs to be permuted for it to be in the right order. However this requires that it is converted to a torch object

        pics_torch = to.from_numpy(pics_np_array) # Converts to torch from numpy array
        # print(type(pics_torch))
        pics = pics_torch.permute(2, 0, 1) # Converted to the right order
        pics_np = pics.numpy()
        
        # We can then loop through the images reshape, rotate and save those of interest

        shp = pics_np.shape[0] # We check the number of pics contained
        Typ[typ] = shp

        if shp == 36:
            for pic_num in range(15, 29):
                image = pics_np[pic_num]
                
                img_height = image.shape[0]
                img_weight = image.shape[1]
                image = np.rot90(image, -1)
                name = pic_name + str(pic_num) + "_" + ".jpeg"
                plt.imsave((savefolder + name), image) # try , cmap="gray"
                # print(image.shape)
                image = cv2.imread((savefolder + name))
                image = cv2.resize(image, dsize=(256, 256), interpolation=cv2.INTER_LINEAR)
                plt.imsave((savefolder + name), image)
                print(image.shape)
                print("succesful resize")
                
        
        elif shp == 256:
            for pic_num in range(100, 160):
                image = pics_np[pic_num]
                img_height = image.shape[0]
                img_weight = image.shape[1]
                image = np.rot90(image, -1)
                name = pic_name + str(pic_num) + "_" + ".jpeg"
                plt.imsave((savefolder + name), image) # try , cmap="gray"
                image = cv2.imread((savefolder + name))
                image = cv2.resize(image, dsize=(256, 256), interpolation=cv2.INTER_LINEAR)
                plt.imsave((savefolder + name), image)
                print(image.shape)
                print("succesful resize")

        else:
            print("file {} shape is {} and does not meet requirements".format(nifti_filename, shp))
            continue

        initial += 1
    except:
        continue

    
print(Typ)
```
